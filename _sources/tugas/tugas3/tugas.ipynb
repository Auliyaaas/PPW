# PTA

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd


def pta():
    data = {"penulis": [], "judul": [], "abstrak_indo": [], "abstrak_english": [],
            "pembimbing_pertama": [], "pembimbing_kedua": [], "prodi": []}

    byprod = 1
    beforeProdiFound = True
    prodiFound = True

    while (prodiFound or beforeProdiFound) and byprod < 100:
        beforeProdiFound = prodiFound

        url = "https://pta.trunojoyo.ac.id/c_search/byprod/{}".format(byprod)
        r = requests.get(url)
        request = r.content
        soup = BeautifulSoup(request, "html.parser")

        prodi_element = soup.select_one('h2:contains("Journal Jurusan")')

        if prodi_element:
            prodi = prodi_element.text.split(' Jurusan ')[1]
            prodiFound = True
        if prodi == '':
            prodiFound = False

        print(f"{byprod}. {prodi}")

        jurnalFound = True
        pageProd = 1

        while jurnalFound and pageProd < 3:
            url = f"https://pta.trunojoyo.ac.id/c_search/byprod/{byprod}/{pageProd}"
            r = requests.get(url)
            request = r.content
            soup = BeautifulSoup(request, "html.parser")

            jurnals = soup.select('li[data-cat="#luxury"]')
            if len(jurnals) < 1:
                jurnalFound = False

            jurnal_index = 0
            while jurnal_index < len(jurnals):
                jurnal = jurnals[jurnal_index]
                try:
                    response = requests.get(jurnal.select_one('a.gray.button')['href'])
                    soup1 = BeautifulSoup(response.content, "html.parser")

                    isi = soup1.select_one('div#content_journal')

                    judul = isi.select_one('a.title').text
                    penulis = isi.select_one('span:contains("Penulis")').text.split(' : ')[1]
                    pembimbing_pertama = isi.select_one('span:contains("Dosen Pembimbing I")').text.split(' : ')[1]
                    pembimbing_kedua = isi.select_one('span:contains("Dosen Pembimbing II")').text.split(' :')[1]

                    abstrak = isi.select('p[align="justify"]')
                    abstrak_indo = abstrak[0].text if len(abstrak) > 0 else "No Indonesian abstract"
                    abstrak_english = abstrak[1].text if len(abstrak) > 1 else "No English abstract"

                    data["penulis"].append(penulis)
                    data["judul"].append(judul)
                    data["pembimbing_pertama"].append(pembimbing_pertama)
                    data["pembimbing_kedua"].append(pembimbing_kedua)
                    data["abstrak_indo"].append(abstrak_indo)
                    data["abstrak_english"].append(abstrak_english)
                    data["prodi"].append(prodi)

                    print(f"  Berhasil crawl: {judul[:50]}...")

                except Exception as e:
                    print(f"  Error processing jurnal {jurnal_index}: {e}")

                jurnal_index += 1

            pageProd += 1

        jurnalFound = True
        byprod += 1

    df = pd.DataFrame(data)
    df.to_csv("pta_test.csv", index=False)

    return df


# === Pemanggilan fungsi & simpan ulang ===
df = pta()
df.to_csv("pta_hasil.csv", index=False, encoding="utf-8-sig")

print("\n✅ Hasil crawling sudah disimpan ke file: pta_hasil.csv")
print(f"\nData berhasil disimpan: {len(df)} records")
print(f"Kolom: {list(df.columns)}")
display(df)
```

# BERITA CNN

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def cnn_rss_scraper():
    kategori_url = {
        "nasional": "https://www.cnnindonesia.com/nasional/rss",
        "internasional": "https://www.cnnindonesia.com/internasional/rss",
        "ekonomi": "https://www.cnnindonesia.com/ekonomi/rss",
        "teknologi": "https://www.cnnindonesia.com/teknologi/rss",
        "olahraga": "https://www.cnnindonesia.com/olahraga/rss",
        "otomotif": "https://www.cnnindonesia.com/otomotif/rss",
        "gaya-hidup": "https://www.cnnindonesia.com/gaya-hidup/rss",
        "hiburan": "https://www.cnnindonesia.com/hiburan/rss"
    }

    # 🚀 Hapus kolom link
    data = {"id": [], "judul": [], "isi_singkat": [], "kategori": []}
    berita_id = 1

    headers = {"User-Agent": "Mozilla/5.0"}

    for kategori, url in kategori_url.items():
        try:
            r = requests.get(url, headers=headers, timeout=10)
            if r.status_code != 200:
                print(f"❌ Gagal akses {url}")
                continue

            soup = BeautifulSoup(r.content, "xml")
            items = soup.find_all("item")

            for item in items:
                judul = item.title.get_text(strip=True) if item.title else ""
                link = item.link.get_text(strip=True) if item.link else ""

                # --- ambil isi singkat dari halaman berita ---
                isi_singkat = ""
                try:
                    r_detail = requests.get(link, headers=headers, timeout=10)
                    if r_detail.status_code == 200:
                        soup_detail = BeautifulSoup(r_detail.content, "html.parser")
                        p = soup_detail.select_one("div.detail-text p")
                        isi_singkat = p.get_text(strip=True) if p else "Tidak ada isi singkat"
                except Exception as e:
                    isi_singkat = f"Gagal ambil isi: {e}"

                # simpan data
                data["id"].append(berita_id)
                data["judul"].append(judul)
                data["isi_singkat"].append(isi_singkat)
                data["kategori"].append(kategori)

                berita_id += 1

        except Exception as e:
            print(f"⚠️ Error scraping {url}: {e}")

    df = pd.DataFrame(data)
    df.to_csv("cnn_berita_rss_full.csv", index=False, encoding="utf-8-sig")
    print(f"\n✅ Data berhasil disimpan: {len(df)} berita")

    # 🚀 Atur tampilan tabel agar rapi
    pd.set_option("display.max_rows", 20)      # tampilkan 20 baris
    pd.set_option("display.max_columns", None) # tampilkan semua kolom
    pd.set_option("display.width", 1000)       # lebar tabel

    return df.head(20)   # tampilkan 20 baris pertama dalam bentuk tabel


# Jalankan fungsi
df_cnn = cnn_rss_scraper()
df_cnn
```

```python
def print_progress(prodi_id, prodi, current_page, total_pages):
    percent = (current_page / total_pages) * 100
    bar_length = 20
    filled_length = int(bar_length * current_page // total_pages)
    bar = '█' * filled_length + '-' * (bar_length - filled_length)
    sys.stdout.write(f'\r[{prodi_id}] {prodi} - Page {current_page}/{total_pages} [{bar}] {percent:.2f}%')
    sys.stdout.flush()
    if current_page == total_pages:
        sys.stdout.write('\n\n')

def pta_links():
    start_time = time.time()  # mulai hitung waktu

    data = {
        "no": [],
        "page": [],
        "link_keluar": []
    }

    no = 1  # nomor urut

    for i in range(1, 42):  # jumlah prodi
        total_pages = 3  # jumlah page
        prodi_name = None

        for j in range(1, total_pages + 1):  # loop page
            url = f"https://pta.trunojoyo.ac.id/c_search/byprod/{i}/{j}"
            r = requests.get(url)
            soup = BeautifulSoup(r.content, "html.parser")
            jurnals = soup.select('li[data-cat="#luxury"]')

            isii = soup.select_one('div#begin')
            if not isii:
                continue
            prodi_full = isii.select_one('h2').text.strip()
            prodi = prodi_full.replace("Journal Jurusan ", "")
            if not prodi_name:
                prodi_name = prodi

            for jurnal in jurnals:
                link = jurnal.select_one('a.gray.button')['href']

                data["no"].append(no)
                data["page"].append(url)          # link page
                data["link_keluar"].append(link)  # link detail
                no += 1

            # update progress bar
            print_progress(i, prodi_name, j, total_pages)

    df = pd.DataFrame(data)
    df.to_csv("pta_links.csv", index=False)

    end_time = time.time()
    elapsed = int(end_time - start_time)
    jam, sisa = divmod(elapsed, 3600)
    menit, detik = divmod(sisa, 60)

    # summary
    print("\n✅ Seluruh link berhasil dikumpulkan!")
    print(f"📊 Total entri: {len(df)}")
    print(f"⏱️ Waktu eksekusi: {jam} jam {menit} menit {detik} detik")

    return df
```
