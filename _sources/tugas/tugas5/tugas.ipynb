```python
import numpy as np

class MyTokenizer:
    def fit_transform(self, texts):
        # Tokenisasi sederhana: lowercase + split
        return [str(text).lower().split() for text in texts]

class MeanEmbeddingVectorizer:
    def __init__(self, word2vec_model):
        self.word2vec = word2vec_model
        # Perbaikan: gunakan vector_size (Gensim ≥ 4.0)
        self.dim = word2vec_model.wv.vector_size

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X_tokenized = MyTokenizer().fit_transform(X)
        embeddings = []
        for words in X_tokenized:
            # Ambil vektor hanya untuk kata yang ada di vocab
            valid_vectors = [
                self.word2vec.wv[word] for word in words
                if word in self.word2vec.wv
            ]
            if valid_vectors:
                embeddings.append(np.mean(valid_vectors, axis=0))
            else:
                embeddings.append(np.zeros(self.dim))
        return np.array(embeddings)

    def fit_transform(self, X, y=None):
        return self.transform(X)
```

```python
import pandas as pd
import re

# Baca dataset
df = pd.read_csv("cnn_tokenized.csv")

clean_txt = []
for w in range(len(df['isi_clean'])):
    # ambil teks dan ubah jadi huruf kecil
    desc = df['isi_clean'][w].lower()

    # hapus tanda baca
    desc = re.sub('[^a-zA-Z]', ' ', desc)

    # hapus tag HTML (kalau ada)
    desc = re.sub("&lt;/?.*?&gt;", " ", desc)

    # hapus angka dan karakter non-alfabet
    desc = re.sub("(\\d|\\W)+", " ", desc)

    clean_txt.append(desc)

# tambahkan kolom baru ke DataFrame
df['clean'] = clean_txt

# cek hasil
print(df[['isi_clean', 'clean']].head())
```

```python
import pandas as pd
import ast
from gensim.models import Word2Vec

# === 1. Baca dataset ===
df = pd.read_csv("cnn_tokenized.csv")

# ubah string list menjadi list Python beneran
df['isi_tokens'] = df['isi_tokens'].apply(ast.literal_eval)

# jadikan list of list (corpus untuk Word2Vec)
corpus = df['isi_tokens'].tolist()

# === 2. Latih Word2Vec ===
model = Word2Vec(
    sentences=corpus,
    vector_size=100,   # ukuran embedding
    window=5,          # konteks
    min_count=2,       # minimal frekuensi kata
    workers=4,         # jumlah thread
    sg=1               # skip-gram (1), CBOW (0)
)

# === 3. Eksplor Embeddings ===
# kata yang mirip dengan 'kpk'
print("Kata mirip dengan 'kpk':")
print(model.wv.most_similar('kpk'))

# analogi: 'indonesia' + 'presiden' - 'jakarta'
print("\nAnalogi:")
print(model.wv.most_similar_cosmul(
    positive=['indonesia', 'presiden'],
    negative=['jakarta']
))

# kata yang tidak nyambung
print("\nKata tidak nyambung:")
print(model.wv.doesnt_match("kpk polisi bali pantai".split()))

# === 4. Simpan embedding ke file ===
filename = 'cnn_embeddings.txt'
model.wv.save_word2vec_format(filename, binary=False)
print(f"\nEmbedding disimpan ke {filename}")
```

```python
mean_embedding_vectorizer = MeanEmbeddingVectorizer(model)
mean_embedded = mean_embedding_vectorizer.fit_transform(df['isi_clean'])
```

```python
df['array']=list(mean_embedded)
```
